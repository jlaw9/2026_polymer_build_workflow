### Display mechanisms templates used in rxn perception

```python
import json
from pathlib import Path

from polymerist.rdutils.reactions.reactions import AnnotatedReaction
from polymerist.rdutils import set_rdkdraw_size
set_rdkdraw_size(300, 3/2)


# RXN_PATH = Path('src/reactions/rxn_smarts.json')
RXN_PATH = Path('src/reactions/rxns_polyID.json')
with RXN_PATH.open('r') as rxnfile:
    rxn_map = json.load(rxnfile)

rxns : dict[str, AnnotatedReaction] = {}
for rxnname, smarts in rxn_map.items():
    rxn = AnnotatedReaction.from_smarts(smarts)
    print(rxnname)
    display(rxn)
    rxns[rxnname] = rxn
```
